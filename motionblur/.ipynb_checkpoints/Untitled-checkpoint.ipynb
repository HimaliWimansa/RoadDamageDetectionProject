```python
import cv2
import numpy as np
import os
import matplotlib.pyplot as plt

# -----------------------------------------
# Folder Setup
# -----------------------------------------
input_folder = "fr"
sharpened_folder = "laplacian_sharpened_frames"

os.makedirs(sharpened_folder, exist_ok=True)

frame_files = sorted(os.listdir(input_folder))

print("\nTotal Files Found in Folder:", len(frame_files))

sharpness_values = []
valid_frames = []

print("\nCalculating Sharpness using Laplacian Variance...\n")

# -----------------------------------------
# STEP 1: Compute Laplacian Variance
# Var( ∇² f(x,y) )
# -----------------------------------------
for file in frame_files:

    if not file.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    path = os.path.join(input_folder, file)
    image = cv2.imread(path)

    if image is None:
        continue

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    laplacian = cv2.Laplacian(gray, cv2.CV_64F)
    lap_var = laplacian.var()

    sharpness_values.append(lap_var)
    valid_frames.append((file, gray, lap_var))

    print(f"{file} → Laplacian Variance: {round(lap_var,2)}")

print("\nTotal Valid Image Frames:", len(valid_frames))

if len(sharpness_values) == 0:
    print("No valid images found.")
    exit()

# -----------------------------------------
# STEP 2: Adaptive Threshold
# Threshold = Mean - Standard Deviation
# -----------------------------------------
mean_sharpness = np.mean(sharpness_values)
std_sharpness = np.std(sharpness_values)
threshold = mean_sharpness - std_sharpness

print("\nMean Sharpness:", round(mean_sharpness,2))
print("Std Deviation :", round(std_sharpness,2))
print("Adaptive Blur Threshold:", round(threshold,2))

# -----------------------------------------
# STEP 3: Classification + Laplacian Sharpening
# g(x,y) = f(x,y) - ∇² f(x,y)
# -----------------------------------------
blur_count = 0
sharp_count = 0
display_limit = 5
display_counter = 0

sharpness_improvements = []

print("\nClassifying Frames...\n")

for file, gray, lap_var in valid_frames:

    if lap_var < threshold:

        print(file, "→ BLURRED")
        blur_count += 1

        # Laplacian Sharpening
        laplacian = cv2.Laplacian(gray, cv2.CV_64F)
        sharpened = gray - laplacian
        sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

        cv2.imwrite(os.path.join(sharpened_folder, file), sharpened)

        # Calculate sharpness AFTER sharpening
        lap_after = cv2.Laplacian(sharpened, cv2.CV_64F)
        sharpness_after = lap_after.var()

        improvement = sharpness_after - lap_var
        sharpness_improvements.append(improvement)

        print("   Sharpness Before :", round(lap_var,2))
        print("   Sharpness After  :", round(sharpness_after,2))
        print("   Improvement      :", round(improvement,2))

        # Edge Strength Evaluation
        edges_before = cv2.Canny(gray, 100, 200)
        edges_after = cv2.Canny(sharpened, 100, 200)

        print("   Edge Pixels Before:", np.count_nonzero(edges_before))
        print("   Edge Pixels After :", np.count_nonzero(edges_after))

        # Display only first 5 blurred frames
        if display_counter < display_limit:

            plt.figure(figsize=(12,4))

            plt.subplot(1,4,1)
            plt.imshow(gray, cmap='gray')
            plt.title("Original")
            plt.axis("off")

            plt.subplot(1,4,2)
            plt.imshow(sharpened, cmap='gray')
            plt.title("Laplacian Sharpened")
            plt.axis("off")

            plt.subplot(1,4,3)
            plt.imshow(edges_before, cmap='gray')
            plt.title("Edges Before")
            plt.axis("off")

            plt.subplot(1,4,4)
            plt.imshow(edges_after, cmap='gray')
            plt.title("Edges After")
            plt.axis("off")

            plt.suptitle(file)
            plt.show()

            display_counter += 1

    else:
        print(file, "→ SHARP (No Processing Needed)")
        sharp_count += 1

# -----------------------------------------
# FINAL SUMMARY
# -----------------------------------------
print("\n========== FINAL SUMMARY ==========")
print("Total Frames Checked :", len(valid_frames))
print("Blurred Frames       :", blur_count)
print("Sharp Frames         :", sharp_count)

if len(sharpness_improvements) > 0:
    avg_improvement = np.mean(sharpness_improvements)
    print("\nAverage Sharpness Improvement:", round(avg_improvement,2))
else:
    print("\nNo blurred frames detected. No improvement calculated.")

print("\nMotion Blur Analysis Completed Successfully.")
```
